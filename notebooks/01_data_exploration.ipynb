```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("../data/raw/student-mat.csv", sep=";")

# Basic exploration
print("Dataset shape:", df.shape)

print("\nFirst 5 rows:")
display(df.head())

print("\nMissing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())

print("\nDataset information:")
df.info()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="G3", bins=20, kde=True)

plt.title("Distribution of Final Mathematics Grades")
plt.xlabel("Final Grade (0–20)")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()
```

```python
# Explore numerical relationships with the final grade
print(df[["age", "studytime", "failures", "absences", "G3"]].corr()["G3"])

# Compare average final grades by study time
print("\nAverage final grade by study time:")
print(df.groupby("studytime")["G3"].mean())

# Visualize study time versus final grade
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="studytime", y="G3")
plt.title("Study Time vs Final Mathematics Grade")
plt.xlabel("Weekly Study Time Category")
plt.ylabel("Final Grade (0–20)")
plt.tight_layout()
plt.show()
```
